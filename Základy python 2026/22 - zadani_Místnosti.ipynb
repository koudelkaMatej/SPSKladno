## Úloha 1 – Třída `Poklad`

Poklad reprezentuje předmět, který může hráč v místnosti najít a sebrat.

### Atributy

| Atribut | Typ | Popis |
|---|---|---|
| `_nazev` | `str` | Název pokladu (chráněný) |
| `_hodnota` | `int` | Hodnota pokladu ve zlatě (≥ 0) |

### Metody / vlastnosti

| Metoda / vlastnost | Popis |
|---|---|
| `__init__(self, nazev, hodnota)` | Nastaví `_nazev`, nastaví `_hodnota` přes property |
| `nazev` (`@property`) | Vrátí `_nazev` (read-only) |
| `hodnota` (`@property`) | Getter: vrátí `_hodnota` |
| `hodnota.setter` | Setter: nastaví jen pokud `hodnota >= 0`, jinak nastaví 0 |
| `__str__(self)` | `"💰 {nazev} ({hodnota} zl.)"` |
| `__repr__(self)` | Vrátí totéž co `__str__` |

<details>
<summary>🔎 Očekávaný výstup testu</summary>

```
💰 Truhla se zlatem (100 zl.)
💰 Rozbitý měšec (0 zl.)
100
```


</details>

````python
# === Úloha 1: Třída Poklad ===


### kód níže NEMĚNIT ###
p1 = Poklad("Truhla se zlatem", 100)
p2 = Poklad("Rozbitý měšec", -10)  # záporná → 0
print(p1)
print(p2)
print(p1.hodnota)
````

## Úloha 2 – Třída `Monstrum` a potomci (dědičnost + polymorfismus)

### Třída `Monstrum`

| Atribut | Typ | Popis |
|---|---|---|
| `_nazev` | `str` | Název monstra (chráněný) |
| `_hp` | `int` | Životy monstra (≥ 0) |
| `_sila` | `int` | Síla útoku (≥ 0) |

| Metoda / vlastnost | Popis |
|---|---|
| `__init__(self, nazev, hp, sila)` | Nastaví `_nazev`, `_hp` a `_sila` přes property |
| `nazev` (`@property`) | Vrátí `_nazev` (read-only) |
| `hp` (`@property` + setter) | ≥ 0, jinak nastaví 0 |
| `sila` (`@property` + setter) | ≥ 0, jinak nastaví 0 |
| `je_nazivu(self)` | Vrátí `True` pokud `hp > 0` |
| `__repr__(self)` | Vrátí totéž co `__str__` |

### Potomci – přepíší **pouze** `zarvi()` a `__init__`

| Třída | `__init__` | `zarvi()` |
|---|---|---|
| `Goblin` | `Goblin()` → `super().__init__("Goblin", 30, 5)` | `"Hehehehe!"` |
| `Skeleton` | `Skeleton()` → `super().__init__("Skeleton", 50, 10)` | `"Klapity-klap!"` |
| `Dragon` | `Dragon()` → `super().__init__("Drak", 200, 40)` | `"🔥 ROOOAAAR!"` |

> 💡 Potomci nemají žádné parametry v `__init__` – vše je pevně dané.

<details>
<summary>🔎 Očekávaný výstup testu</summary>

```
👾 Goblin (HP: 30, Síla: 5)
Hehehehe!
👾 Skeleton (HP: 50, Síla: 10)
Klapity-klap!
👾 Drak (HP: 200, Síla: 40)
🔥 ROOOAAAR!
True
[👾 Goblin (HP: 30, Síla: 5), 👾 Drak (HP: 200, Síla: 40)]
```

</details>


</details>

````python
# === Úloha 2: Monstrum + Goblin, Skeleton, Dragon ===


### kód níže NEMĚNIT ###
g = Goblin()
s = Skeleton()
d = Dragon()

print(g)
print(g.zarvi())
print(s)
print(s.zarvi())
print(d)
print(d.zarvi())
print(d.je_nazivu())
print([g, d])  # ověří __repr__
````

## Úloha 3 – Třída `Mistnost` (kompozice)

Místnost je **uzel grafové mapy**. Obsahuje název, volitelný obsah (Poklad nebo Monstrum) a seznam východů (jiné místnosti).

### Atributy

| Atribut | Typ | Popis |
|---|---|---|
| `_nazev` | `str` | Název místnosti |
| `_popis` | `str` | Krátký popis (1 věta) |
| `_obsah` | `Poklad \| Monstrum \| None` | Poklad nebo monstrum (nebo nic) |
| `_exity` | `list[Mistnost]` | Seznam propojených místností (začíná prázdný) |

### Metody / vlastnosti

| Metoda / vlastnost | Popis |
|---|---|
| `__init__(self, nazev, popis, obsah=None)` | Nastaví atributy, `_exity` = prázdný seznam |
| `nazev` (`@property`) | Vrátí `_nazev` (read-only) |
| `popis` (`@property`) | Vrátí `_popis` (read-only) |
| `obsah` (`@property`) | Vrátí `_obsah` |
| `obsah.setter` | Nastaví `_obsah` (pro odebrání pokladu – nastavíme na `None`) |
| `exity` (`@property`) | Vrátí **kopii** seznamu `_exity` (ochrana dat) |
| `pridej_exit(self, mistnost)` | Přidá místnost do `_exity` (oboustranně – přidá i zpětný exit) |
| `__str__(self)` | Viz ukázka níže |
| `__repr__(self)` | Vrátí `_nazev` |

### Formát `__str__`

```
📍 Vstupní jeskyně
   Tmavý vlhký vstup do podzemí.
   Obsah: 💰 Stará mince (5 zl.)
   Východy: Chodba, Jezírko
```

Pokud místnost nemá obsah: `Obsah: (prázdno)`

> 💡 `pridej_exit` přidá propojení **oboustranně** – abyste nemuseli psát dvakrát:
> ```python
> m1.pridej_exit(m2)  # automaticky přidá i m2 → m1
> ```

<details>
<summary>🔎 Očekávaný výstup testu</summary>

```
📍 Vstupní jeskyně
   Tmavý vlhký vstup do podzemí.
   Obsah: 💰 Stará mince (5 zl.)
   Východy: Chodba

📍 Chodba
   Úzká kamenná chodba.
   Obsah: 👾 Goblin (HP: 30, Síla: 5)
   Východy: Vstupní jeskyně
```

</details>


````python
# === Úloha 3: Třída Mistnost ===


### kód níže NEMĚNIT ###
m1 = Mistnost("Vstupní jeskyně", "Tmavý vlhký vstup do podzemí.", Poklad("Stará mince", 5))
m2 = Mistnost("Chodba", "Úzká kamenná chodba.", Goblin())

m1.pridej_exit(m2)  # oboustranně!

print(m1)
print()
print(m2)
````

## Úloha 4 – Třída `Hrac`

Hráč má jméno, životy, zlato a nachází se v určité místnosti.

### Atributy

| Atribut | Typ | Popis |
|---|---|---|
| `_jmeno` | `str` | Jméno hráče |
| `_hp` | `int` | Životy (≥ 0), výchozí 100 |
| `_zlato` | `int` | Nasbírané zlato, výchozí 0 |
| `_pozice` | `Mistnost` | Aktuální místnost |

### Metody / vlastnosti

| Metoda / vlastnost | Popis |
|---|---|
| `__init__(self, jmeno, startovni_mistnost)` | Nastaví jméno, `_hp=100`, `_zlato=0`, `_pozice` |
| `jmeno` (`@property`) | read-only |
| `hp` (`@property` + setter) | ≥ 0, jinak 0 |
| `zlato` (`@property` + setter) | ≥ 0, jinak 0 |
| `pozice` (`@property` + setter) | Aktuální místnost |
| `je_nazivu(self)` | `True` pokud `hp > 0` |
| `seber_poklad(self)` | Pokud `_pozice.obsah` je `Poklad`: přičte hodnotu ke zlatu, nastaví obsah místnosti na `None`, vrátí text `"Sebral jsi {nazev} (+{hodnota} zl.)"`. Jinak vrátí `"Nic tu není."` |
| `jdi(self, index)` | Přesune hráče do místnosti `_pozice.exity[index]`. Vrátí `str(nová pozice)`. Pokud index neexistuje, vrátí `"Neplatný východ!"` |
| `__str__(self)` | `"🧝 {jmeno} | ❤️ {hp} HP | 💰 {zlato} zl. | 📍 {pozice.nazev}"` |

> 💡 Pro kontrolu typu použijte `isinstance(objekt, Trida)`:
> ```python
> if isinstance(self._pozice.obsah, Poklad):
> ```

<details>
<summary>🔎 Očekávaný výstup testu</summary>

```
🧝 Hrdina | ❤️ 100 HP | 💰 0 zl. | 📍 Vstupní jeskyně
Sebral jsi Stará mince (+5 zl.)
🧝 Hrdina | ❤️ 100 HP | 💰 5 zl. | 📍 Vstupní jeskyně
Nic tu není.
```

</details>


````python
# === Úloha 4: Třída Hrac ===


### kód níže NEMĚNIT ###
# Znovu vytvoříme mapu (m1 s pokladem)
m1 = Mistnost("Vstupní jeskyně", "Tmavý vlhký vstup do podzemí.", Poklad("Stará mince", 5))
m2 = Mistnost("Chodba", "Úzká kamenná chodba.", Goblin())
m1.pridej_exit(m2)

hrac = Hrac("Hrdina", m1)
print(hrac)
print(hrac.seber_poklad())
print(hrac)
print(hrac.seber_poklad())  # podruhé – už nic
````

## Úloha 5 – Metoda `bojuj` ve třídě `Hrac`

Přidejte do třídy `Hrac` metodu `bojuj(self)`:

### Logika souboje

1. Zkontroluje, zda je v aktuální místnosti `Monstrum` (pomocí `isinstance`)
2. Pokud ne → vrátí `"Žádný nepřítel na dohled."`
3. Pokud ano, probíhá **tahový souboj** (while cyklus):
   - **Hráč útočí** – monstrum ztratí **20 HP** (pevná síla hráče)
   - Pokud monstrum padlo (`hp <= 0`) → odstraní ho z místnosti (`obsah = None`), vrátí `"⚔️ Porazil jsi {nazev}!"`
   - **Monstrum útočí** – hráč ztratí HP podle `sila` monstra
   - Pokud hráč padl → vrátí `"💀 {nazev} tě porazil..."`
4. Každé kolo vypíše: `"  Hráč ❤️ {hrac_hp} vs {monstrum_nazev} ❤️ {monstrum_hp}"`

<details>
<summary>🔎 Očekávaný výstup testu</summary>

```
📍 Chodba
   Úzká kamenná chodba.
   Obsah: 👾 Goblin (HP: 30, Síla: 5)
   Východy: Vstupní jeskyně

  Hráč ❤️ 100 vs Goblin ❤️ 10
  Hráč ❤️ 95 vs Goblin ❤️ -10
⚔️ Porazil jsi Goblin!

📍 Chodba
   Úzká kamenná chodba.
   Obsah: (prázdno)
   Východy: Vstupní jeskyně
```

</details>


````python
# === Úloha 5: Metoda bojuj ===
# Přidejte metodu bojuj() do třídy Hrac výše a spusťte tento test.

### kód níže NEMĚNIT ###
# Přesun do chodby (tam je Goblin)
hrac.jdi(0)
print(hrac.pozice)
print()
print(hrac.bojuj())
print()
print(hrac.pozice)  # Goblin zmizí
````

### Mapa

```

                          ┌───────┼───────┐

### Místnosti a jejich obsah

- Vstup ↔ Chodba

## 📋 Shrnutí – co jste si procvičili

| Koncept | Kde v úloze |
|---|---|
| `class`, `__init__`, `self` | Všechny třídy |
| `__str__`, `__repr__` | `Poklad`, `Monstrum`, `Mistnost`, `Hrac` |
| `@property` (read-only) | `Poklad.nazev`, `Monstrum.nazev`, `Mistnost.nazev`, `Hrac.jmeno` |
| `@property` + setter (validace) | `Poklad.hodnota`, `Monstrum.hp/.sila`, `Hrac.hp/.zlato` |
| Dědičnost (`super()`) | `Goblin/Skeleton/Dragon(Monstrum)` |
| Polymorfismus | `zarvi()` – každé monstrum řve jinak |
| Kompozice (HAS-A) | `Mistnost` obsahuje `Poklad`/`Monstrum`, `Mistnost` má seznam `Mistnost` |
| `isinstance()` | `Hrac.seber_poklad()`, `Hrac.bojuj()` |
| Seznamy objektů | `_exity`, průchod mapou |
| While cyklus + OOP | Soubojový systém, herní smyčka |
