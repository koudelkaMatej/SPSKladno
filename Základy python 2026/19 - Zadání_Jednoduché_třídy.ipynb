# 15 – Zadání: Jednoduché třídy – Python styl vs. Java styl

V tomto notebooku si procvičíte dva přístupy k řízení přístupu k atributům:

| | **Python styl** (`@property`) | **Java styl** (`get/set` metody) |
|:---|:---|:---|
| **Čtení** | `osoba.jmeno` | `osoba.get_jmeno()` |
| **Zápis** | `osoba.jmeno = "Petr"` | `osoba.set_jmeno("Petr")` |
| **Validace** | V `@jmeno.setter` | V `set_jmeno()` |
| **Výhoda** | Přirozený zápis, Pythonic | Jasně viditelné, že jde o metodu |

> 🎯 **Cíl:** Naučit se oba přístupy a pochopit, proč Python preferuje `@property`. Obě části mají **stejné zadání**, abyste viděli rozdíl v implementaci.

# 🔀 Srovnání obou stylů

Podívejte se, jak se liší **použití** obou přístupů na stejném příkladu:

```python
# === Python styl (@property) ===
karel = Clovek("Karel", "Novák", 30, "123456789", "karel@test.cz", vzp)
karel.vek = 45              # setter se zavolá automaticky
print(karel.vek)             # getter se zavolá automaticky
karel.telefon = "111222333"  # validace proběhne sama
print(karel)

# === Java styl (get/set) ===
karel = ClovekJava("Karel", "Novák", 30, "123456789", "karel@test.cz", vzp)
karel.set_vek(45)            # musíme explicitně zavolat metodu
print(karel.get_vek())       # musíme explicitně zavolat metodu
karel.set_telefon("111222333")
print(karel)
```

| Vlastnost | Python styl (`@property`) | Java styl (`get/set`) |
|:---|:---|:---|
| **Zápis hodnoty** | `osoba.vek = 45` | `osoba.set_vek(45)` |
| **Čtení hodnoty** | `osoba.vek` | `osoba.get_vek()` |
| **Validace** | Skrytá v setteru – uživatel nevidí | Viditelná – uživatel volá metodu |
| **Přirozenost v Pythonu** | ✅ Pythonic, elegantní | ❌ Nepřirozené, "ukecaný" kód |
| **Kompatibilita** | Kód se nemění, i když přidáme validaci | Musíme přepsat všechna volání |
| **Použití v praxi** | Python, C#, Kotlin | Java, C++ |

> 🎯 **Závěr:** V Pythonu **vždy používejte `@property`**. Java styl s `get/set` metodami je tu pouze pro pochopení konceptu, který potkáte v jiných jazycích.
