# 16 – Zadání: Útulek 🐾

Procvičení **dědičnosti**, **polymorfismu** a **kompozice** na příkladu zvířecího útulku.

Úlohy na sebe navazují – každá další rozšiřuje předchozí kód.

```
          ┌─────────────┐
          │   Zvire     │   ← rodičovská třída
          │─────────────│
          │ jmeno       │
          │ hmotnost    │
          │─────────────│
          │ zarvi()     │
          │ __str__()   │
          └────┬────────┘
       ┌───────┼───────┐
  ┌────┴───┐ ┌┴─────┐ ┌┴────────┐
  │  Pes   │ │Kocka │ │Papousek │  ← potomci
  │────────│ │──────│ │─────────│
  │plemeno │ │barva │ │slovnik  │
  │────────│ │──────│ │─────────│
  │zarvi() │ │zarvi │ │zarvi()  │
  └────────┘ └──────┘ └─────────┘
                 │
          ┌──────┴──────┐
          │   Utulek    │   ← kompozice (MÁ zvířata)
          │─────────────│
          │ nazev       │
          │ kapacita    │
          │ zvirata[]   │
          │─────────────│
          │ pridej()    │
          │ odeber()    │
          │ vypis()     │
          └─────────────┘
```

> **Předpoklady:** Třídy, `__init__`, `__str__`, property, dědičnost (`class Potomek(Rodic)`), `super()`, přepisování metod, polymorfismus.

## Úloha 1 – Rodičovská třída `Zvire`

Vytvořte třídu `Zvire`, která bude mít:

| Atribut | Typ | Poznámka |
|:--------|:----|:---------|
| `jmeno` | str | veřejný |
| `hmotnost` | float | **property** – nesmí být záporná (minimum 0) |

| Metoda | Popis |
|:-------|:------|
| `zarvi()` | Vypíše `"..."` (obecné zvíře neví, jak řvát) |
| `__str__()` | Vrátí `"{jméno} ({hmotnost} kg)"` |

> ⚠️ **Kód pod čarou neměňte!**

<details>
<summary>🔎 Očekávaný výstup</summary>

```
Anička (5.0 kg)
...
Hmotnost po pokusu nastavit -3: 0
```

</details>

````python
# Úloha 1: Váš kód sem 👇



### kód níže NEMĚNIT ###
z = Zvire("Anička", 5.0)
print(z)
z.zarvi()
z.hmotnost = -3
print(f"Hmotnost po pokusu nastavit -3: {z.hmotnost}")
````

## Úloha 2 – Potomci: `Pes` a `Kocka`

Vytvořte dvě třídy, které **dědí** od `Zvire`:

### Třída `Pes(Zvire)`

| Atribut | Typ | Poznámka |
|:--------|:----|:---------|
| `plemeno` | str | veřejný, nový atribut |

| Metoda | Popis |
|:-------|:------|
| `zarvi()` | Vypíše `"Haf haf!"` **(přepsání)** |
| `__str__()` | Vrátí `"Pes {jméno} ({hmotnost} kg) – {plemeno}"` |

### Třída `Kocka(Zvire)`

| Atribut | Typ | Poznámka |
|:--------|:----|:---------|
| `barva` | str | veřejný, nový atribut |

| Metoda | Popis |
|:-------|:------|
| `zarvi()` | Vypíše `"Mňau!"` **(přepsání)** |
| `__str__()` | Vrátí `"Kočka {jméno} ({hmotnost} kg) – {barva}"` |

> 💡 **Nezapomeňte** v `__init__` zavolat `super().__init__(jmeno, hmotnost)`, aby potomci zdědili atributy rodiče.

> ⚠️ **Kód pod čarou neměňte!**

<details>
<summary>🔎 Očekávaný výstup</summary>

```
Pes Rex (25.0 kg) – labrador
Haf haf!
Kočka Míca (4.0 kg) – černá
Mňau!
Je Rex zvíře? True
Je Míca pes? False
```

</details>

````python
# Úloha 2: Váš kód sem 👇



### kód níže NEMĚNIT ###
rex = Pes("Rex", 25.0, "labrador")
print(rex)
rex.zarvi()

mica = Kocka("Míca", 4.0, "černá")
print(mica)
mica.zarvi()

print(f"Je Rex zvíře? {isinstance(rex, Zvire)}")
print(f"Je Míca pes? {isinstance(mica, Pes)}")
````

## Úloha 4 – Třída `Utulek`

Vytvořte třídu `Utulek`, která **obsahuje** (kompozice) seznam zvířat:

| Atribut | Typ | Poznámka |
|:--------|:----|:---------|
| `nazev` | str | veřejný |
| `kapacita` | int | **property** – maximální počet zvířat, minimum 1 |
| `_zvirata` | list | **skrytý/ protected** – seznam zvířat (začíná prázdný) |

| Metoda | Popis |
|:-------|:------|
| `pocet_zvirat` | **property (read-only)** – vrátí aktuální počet zvířat |
| `pridej(zvire)` | Přidá zvíře do seznamu. Pokud je útulek plný, vypíše `"Útulek {název} je plný! Nelze přidat {jméno zvířete}."` |
| `odeber(jmeno)` | Odebere zvíře podle jména. Pokud není nalezeno, vypíše `"Zvíře {jméno} nebylo nalezeno."` |
| `__str__()` | Vrátí přehled útulku (viz ukázka níže) |

> 💡 **Metoda `pridej()` by měla přijímat pouze instance třídy `Zvire`** (nebo jejích potomků). Pokud dostane něco jiného, vypíše `"Lze přidat pouze zvíře!"`.

> ⚠️ **Kód pod čarou neměňte!**

<details>
<summary>🔎 Očekávaný výstup</summary>

```
=== Útulek Naděje (0/3) ===
Žádná zvířata.

=== Útulek Naděje (3/3) ===
  1. Pes Rex (25.0 kg) – labrador
  2. Kočka Míca (4.0 kg) – černá
  3. Pes Bára (12.0 kg) – pudl

Útulek Naděje je plný! Nelze přidat Anička.

=== Útulek Naděje (2/3) ===
  1. Pes Rex (25.0 kg) – labrador
  2. Pes Bára (12.0 kg) – pudl

Zvíře Nemo nebylo nalezeno.
```

</details>

````python
# Úloha 4: Váš kód sem 👇



### kód níže NEMĚNIT ###
utulek = Utulek("Naděje", 3)
print(utulek)

utulek.pridej(Pes("Rex", 25.0, "labrador"))
utulek.pridej(Kocka("Míca", 4.0, "černá"))
utulek.pridej(Pes("Bára", 12.0, "pudl"))
print(utulek)

utulek.pridej(Zvire("Anička", 5.0))  # útulek je plný

utulek.odeber("Míca")
print(utulek)

utulek.odeber("Nemo")  # neexistuje
````

Přidejte do třídy `Utulek` metodu:

> 🎯 **Tady krásně vidíte polymorfismus:** Útulek nemusí vědět, jestli má psa nebo kočku – prostě zavolá `zarvi()` a každé zvíře zareaguje po svém.

> ⚠️ **Kód pod čarou neměňte!**

<details>
<summary>🔎 Očekávaný výstup</summary>

```
🔊 Zvířata v útulku Naděje řvou:
  Rex: Haf haf!
  Míca: Mňau!
  Bára: Haf haf!
```

</details>

````python
# Úloha 5: Přidejte metodu vsichni_zarvi() do třídy Utulek 👇



### kód níže NEMĚNIT ###
utulek = Utulek("Naděje", 5)
utulek.pridej(Pes("Rex", 25.0, "labrador"))
utulek.pridej(Kocka("Míca", 4.0, "černá"))
utulek.pridej(Pes("Bára", 12.0, "pudl"))
utulek.vsichni_zarvi()
````

## Úloha 6 – Nový potomek: `Papousek`

Vytvořte třídu `Papousek(Zvire)`:

| Atribut | Typ | Poznámka |
|:--------|:----|:---------|
| `slovnik` | list[str] | Seznam slov, která papoušek umí (výchozí prázdný seznam) |

| Metoda | Popis |
|:-------|:------|
| `zarvi()` | Vypíše `"Kráá!"` |
| `nauc_slovo(slovo)` | Přidá slovo do slovníku |
| `mluv()` | Vypíše náhodné slovo ze slovníku. Pokud je slovník prázdný, vypíše `"..."` |
| `__str__()` | Vrátí `"Papoušek {jméno} ({hmotnost} kg) – zná {počet} slov"` |

> 💡 Pro náhodný výběr použijte `import random` a `random.choice(seznam)`.

> ⚠️ **Kód pod čarou neměňte!** (Výstup `mluv()` bude náhodný – ověřte vizuálně.)

<details>
<summary>🔎 Očekávaný výstup (příklad)</summary>

```
Papoušek Pepa (0.5 kg) – zná 0 slov
Kráá!
...
Papoušek Pepa (0.5 kg) – zná 3 slov
Ahoj
```

</details>

````python
# Úloha 6: Váš kód sem 👇



### kód níže NEMĚNIT ###
pepa = Papousek("Pepa", 0.5)
print(pepa)
pepa.zarvi()
pepa.mluv()             # slovník prázdný → "..."

pepa.nauc_slovo("Ahoj")
pepa.nauc_slovo("Pepa")
pepa.nauc_slovo("Kráska")
print(pepa)
pepa.mluv()             # náhodné slovo
````

## 📋 Co jste si procvičili

| Koncept | Kde v úlohách |
|:--------|:--------------|
| **Rodičovská třída** | `Zvire` (úloha 1) |
| **Dědičnost** (`class Potomek(Rodic)`) | `Pes`, `Kocka`, `Papousek` (úlohy 2, 6) |
| **`super().__init__()`** | Ve všech potomcích |
| **Přepisování metod** (override) | `zarvi()`, `__str__()` v každém potomkovi |
| **Polymorfismus** | Procházení seznamu zvířat, `vsichni_zarvi()` (úlohy 3, 5) |
| **Kompozice** (vztah MÁ) | `Utulek` obsahuje seznam `Zvire` (úloha 4) |
| **`isinstance()`** | Kontrola typu v `pridej()` a v bonusu |
| **Property** | `hmotnost` v `Zvire`, `kapacita` v `Utulek` |

> 🎯 **Klíčový poznatek:** Útulek **MÁ** zvířata (kompozice), ale Pes **JE** Zvíře (dědičnost). Proto `Utulek` nedědi od `Zvire`, ale obsahuje seznam zvířat jako atribut.
