```
┌─────────────────────┐
│      Polozka        │

│ _polozky : list     │
│─────────────────────│
│ pridej(polozka)     │
│ odeber(nazev)       │
│ celkova_cena        │  ← read-only property

└─────────────────────┘
```

## Úloha 1 – Třída `Polozka`

Vytvořte třídu `Polozka`, která reprezentuje jednu položku v nákupním košíku:

| Atribut | Typ | Poznámka |
|:--------|:----|:---------|
| `nazev` | str | veřejný, bez property |
| `cena` | float | **property** – cena za kus, nesmí být záporná (minimum 0) |
| `pocet` | int | **property** – počet kusů, minimum **1** |

| Metoda / Property | Popis |
|:------------------|:------|
| `celkova_cena` | **read-only property** – vrátí `cena × pocet` |
| `__str__()` | Vrátí `"{nazev}: {pocet}x {cena} Kč = {celkova_cena} Kč"` |
| `__repr__()` | Vrátí totéž co `__str__()` (aby fungoval výpis v seznamu) |

> ⚠️ **Kód pod čarou neměňte!**

<details>
<summary>🔎 Očekávaný výstup</summary>

```
Rohlík: 10x 3.5 Kč = 35.0 Kč
Cena po úpravě na -5: 0
Počet po úpravě na 0: 1
Celková cena: 0
Máslo: 1x 45.0 Kč = 45.0 Kč
```

</details>

````python
# Úloha 1: Váš kód sem 👇



### kód níže NEMĚNIT ###
rohlik = Polozka("Rohlík", 3.5, 10)
print(rohlik)
rohlik.cena = -5
print(f"Cena po úpravě na -5: {rohlik.cena}")
rohlik.pocet = 0
print(f"Počet po úpravě na 0: {rohlik.pocet}")
print(f"Celková cena: {rohlik.celkova_cena}")
maslo = Polozka("Máslo", 45.0)
print(maslo)
````

## Úloha 3 – Třída `NakupniKosik`

Vytvořte třídu `NakupniKosik`, která **obsahuje** (kompozice) seznam položek:

| Atribut | Typ | Poznámka |
|:--------|:----|:---------|
| `nazev` | str | veřejný (např. "Lidl", "Albert") |
| `budget` | float | **property** – kolik mám peněz, minimum 0 |
| `_polozky` | list | **skrytý** – seznam položek (začíná prázdný) |

| Metoda / Property | Popis |
|:------------------|:------|
| `pocet_polozek` | **read-only property** – počet položek v košíku |
| `celkova_cena` | **read-only property** – součet `celkova_cena` všech položek |
| `zbyva_penez` | **read-only property** – `budget - celkova_cena` |
| `pridej(polozka)` | Přidá položku. Pokud celková cena po přidání přesáhne budget, vypíše `"⚠️ Nedostatek peněz! {nazev} ({celkova_cena} Kč) se nevešlo do rozpočtu."` a **nepřidá ji**. Přijímá pouze instance `Polozka`. |
| `odeber(nazev)` | Odebere položku podle názvu. Pokud nenalezena, vypíše `"Položka '{nazev}' nebyla nalezena."` |
| `__str__()` | Vrátí přehled košíku (viz ukázka níže) |

> ⚠️ **Kód pod čarou neměňte!**

<details>
<summary>🔎 Očekávaný výstup</summary>

```
🛒 Košík Lidl (0 položek) – Budget: 200.0 Kč
  Košík je prázdný.
  Zbývá: 200.0 Kč

🛒 Košík Lidl (3 položek) – Budget: 200.0 Kč
  1. Rohlík: 10x 3.5 Kč = 35.0 Kč
  2. Máslo: 1x 45.0 Kč = 45.0 Kč
  3. Mléko: 2x 22.9 Kč = 45.8 Kč
  Celkem: 125.8 Kč | Zbývá: 74.2 Kč

⚠️ Nedostatek peněz! Televize (15000.0 Kč) se nevešlo do rozpočtu.

Položka 'Banán' nebyla nalezena.

🛒 Košík Lidl (2 položek) – Budget: 200.0 Kč
  1. Rohlík: 10x 3.5 Kč = 35.0 Kč
  2. Mléko: 2x 22.9 Kč = 45.8 Kč
  Celkem: 80.8 Kč | Zbývá: 119.2 Kč
```

</details>

````python
# Úloha 3: Váš kód sem 👇



### kód níže NEMĚNIT ###
kosik = NakupniKosik("Lidl", 200.0)
print(kosik)

kosik.pridej(Polozka("Rohlík", 3.5, 10))
kosik.pridej(Polozka("Máslo", 45.0))
kosik.pridej(Polozka("Mléko", 22.9, 2))
print(kosik)

kosik.pridej(Polozka("Televize", 15000.0))  # přes budget

kosik.odeber("Banán")  # neexistuje
kosik.odeber("Máslo")
print(kosik)
````

## Úloha 4 – Změna počtu kusů

Přidejte do třídy `NakupniKosik` metodu:

| Metoda | Popis |
|:-------|:------|
| `zmen_pocet(nazev, novy_pocet)` | Najde položku podle názvu a změní její počet. Pokud by nová celková cena košíku přesáhla budget, změnu **neprovede** a vypíše `"⚠️ Změna počtu by překročila budget!"`. Pokud položka neexistuje, vypíše `"Položka '{nazev}' nebyla nalezena."` |

> 💡 **Tip:** Uložte si starý počet, změňte, zkontrolujte celkovou cenu košíku – pokud přesahuje budget, vraťte starý počet zpět.

> ⚠️ **Kód pod čarou neměňte!**

<details>
<summary>🔎 Očekávaný výstup</summary>

```
🛒 Košík Albert (3 položek) – Budget: 100.0 Kč
  1. Chleba: 1x 32.0 Kč = 32.0 Kč
  2. Šunka: 1x 25.0 Kč = 25.0 Kč
  3. Sýr: 1x 35.0 Kč = 35.0 Kč
  Celkem: 92.0 Kč | Zbývá: 8.0 Kč

⚠️ Změna počtu by překročila budget!

🛒 Košík Albert (3 položek) – Budget: 100.0 Kč
  1. Chleba: 2x 32.0 Kč = 64.0 Kč
  2. Šunka: 1x 25.0 Kč = 25.0 Kč
  3. Sýr: 1x 35.0 Kč = 35.0 Kč
  Celkem: 124.0 Kč | Zbývá: -24.0 Kč
```
Počkat – to nejde! Přečtěte si znovu zadání. 😉

Správný výstup po pokusu o změnu chleba na 2x:
```
⚠️ Změna počtu by překročila budget!
```
Košík zůstane nezměněný.

</details>

````python
# Úloha 4: Přidejte metodu zmen_pocet() do třídy NakupniKosik 👇



### kód níže NEMĚNIT ###
kosik2 = NakupniKosik("Albert", 100.0)
kosik2.pridej(Polozka("Chleba", 32.0))
kosik2.pridej(Polozka("Šunka", 25.0))
kosik2.pridej(Polozka("Sýr", 35.0))
print(kosik2)

kosik2.zmen_pocet("Chleba", 5)   # 5× chleba = 160 Kč → přes budget
kosik2.zmen_pocet("Chleba", 2)   # 2× chleba = 64 + 25 + 35 = 124 → přes budget
print(kosik2)
````

## Úloha 5 – Účtenka

Přidejte do třídy `NakupniKosik` metodu:

| Metoda | Popis |
|:-------|:------|
| `vytiskni_uctenku()` | Vypíše formátovanou účtenku (viz ukázka). Pokud je košík prázdný, vypíše `"Košík je prázdný, nelze vytisknout účtenku."` |

<details>
<summary>🔎 Očekávaný výstup</summary>

```
══════════════════════════════════

══════════════════════════════════

──────────────────────────────────

```

</details>

> 💡 **Tip:** Pro formátování použijte f-stringy s šířkou: `f"{text:<15}"` (zarovná vlevo na 15 znaků), `f"{cislo:>8.1f}"` (zarovná vpravo na 8 znaků s 1 des. místem).

````python
# Úloha 5: Přidejte metodu vytiskni_uctenku() do třídy NakupniKosik 👇

### kód níže NEMĚNIT ###
kosik3 = NakupniKosik("Lidl", 200.0)
kosik3.pridej(Polozka("Rohlík", 3.5, 10))
kosik3.pridej(Polozka("Máslo", 45.0))
kosik3.pridej(Polozka("Mléko", 22.9, 2))
kosik3.vytiskni_uctenku()
````

## 📋 Co jste si procvičili

| Koncept | Kde v úlohách |
|:--------|:--------------|
| **Třída s property** | `Polozka` – cena ≥ 0, počet ≥ 1 (úloha 1) |
| **Read-only property** | `celkova_cena`, `zbyva_penez`, `pocet_polozek` (úlohy 1, 3) |
| **`__str__` a `__repr__`** | `Polozka` – výpis jednotlivě i v seznamu (úlohy 1, 2) |
| **Kompozice** (vztah MÁ) | `NakupniKosik` obsahuje seznam `Polozka` (úloha 3) |
| **Validace v metodách** | `pridej()`, `zmen_pocet()` – kontrola budgetu (úlohy 3, 4) |
| **Formátování f-stringy** | `vytiskni_uctenku()` (úloha 5) |
| **`isinstance()`** | Kontrola typu v `pridej()` (úloha 3) |

> 🎯 **Klíčový poznatek:** Košík **MÁ** položky – to je **kompozice**. Položky ukládáme do seznamu jako atribut třídy a pracujeme s nimi přes metody košíku.
